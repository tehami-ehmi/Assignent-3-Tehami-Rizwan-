# Q1. Nested Dictionary - Update Stock & Add Product

```python
products = {
    101: {"name": "Laptop", "category": "Electronics", "price": 1200, "stock": 50},
    102: {"name": "Shirt", "category": "Apparel", "price": 25, "stock": 200},
    103: {"name": "Coffee Maker", "category": "Home Appliances", "price": 80, "stock": 30}
}
products[102]["stock"] = 250
print(products)
products[104] = {"name": "Smartphone", "category": "Electronics", "price": 800, "stock": 100}
print(products)
```

# Q2. Age Group Classification

```python
age = int(input("Enter age: "))

if 0 <= age <= 12:
    print("Child")
elif 13 <= age <= 19:
    print("Teenager")
elif 20 <= age <= 64:
    print("Adult")
elif age >= 65:
    print("Senior")
else:
    print("Invalid age")
```

# Q3. Book Details in a Tuple

```python
title = input("Enter title of the book: ")
author = input("Enter author's name: ")
year = input("Enter year of publication: ")
price = float(input("Enter price of the book: "))

book = (title, author, year, price)

print("Title:", book[0])
print("Author:", book[1])
print("Year of Publication:", book[2])
print("Price: $" + str(book[3]))
```

# Q4. Positive / Negative / Zero

# Q7. Add & Remove Student

```python
student_details = {
    "Ali": {"age": 20, "grade": "A", "city": "Karachi"},
    "Sara": {"age": 21, "grade": "B", "city": "Lahore"},
    "Omar": {"age": 19, "grade": "A", "city": "Islamabad"},
    "Hina": {"age": 22, "grade": "C", "city": "Faisalabad"}
}
student_details["Zara"] = {"age": 20, "grade": "B", "city": "Multan"}
student_details.pop("Omar")
print(student_details)
```

# Q8. Inventory Management

```python
inventory = {
    "The Great Gatsby": {"price": 10.99, "quantity": 5},
    "Rich dad poor dad": {"price": 8.99, "quantity": 8},
    "Harry Potter": {"price": 12.50, "quantity": 2},
}
inventory.update({"Atomic Habits": {"price": 15.00, "quantity": 4}})
titles = inventory.keys()
print("Book Titles:", list(titles))
for book in list(inventory.keys()):
    if inventory[book]["quantity"] == 0:
        inventory.pop(book)
print(inventory)
```

# Q10. Vowel / Consonant / Invalid

```python
ch = input("Enter a single letter: ")

if len(ch) == 1 and ch.isalpha():
    if ch.lower() in "aeiou":
        print(f"{ch} is a Vowel")
    else:
        print(f"{ch} is a Consonant")
else:
    print("Please enter valid alphabet")
```
